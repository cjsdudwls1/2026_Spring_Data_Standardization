```python
import seaborn as sns
```

```python
# 문제 1. 데이터 준비 및 탐색
penguins = sns.load_dataset('penguins')
penguins_clean = penguins.dropna()
gentoo_flipper = penguins_clean[penguins_clean['species'] == 'Gentoo']['flipper_length_mm']
population_mean = gentoo_flipper.mean()

print(f"문제 1. 모평균(전체 평균): {population_mean}")
```

```python
# 문제 2. 표본 추출 및 점추정
sample_5_mean = gentoo_flipper.sample(n=5, random_state=42).mean()
sample_50_mean = gentoo_flipper.sample(n=50, random_state=42).mean()

print(f"문제 2. 5개 표본의 표본평균: {sample_5_mean}")
print(f"문제 2. 50개 표본의 표본평균: {sample_50_mean}")
print("결과 및 이유: 50개의 표본을 추출한 표본평균이 모평균에 더 근접합니다. 대수의 법칙(Law of Large Numbers)에 따라 표본의 크기가 커질수록 표본평균의 오차가 줄어들어 모평균에 더 수렴하기 때문입니다.")
```
